```python
som_1=0
som_2 = 0
coeff_1=0
coeff_2 = 0

for i in range (101,116):
    note=float(input(f"indiquer la note obtenue dans la ressource R{i}"))
    coeff=float(input(f"indiquer le coefficient pour la ressource R{i}"))
    som_1 += note*coeff
    coeff_1 += coeff


for i in range (11,17):
    note=float(input(f"indiquer la note obtenue dans la ressource SAE{i}"))
    coeff=float(input(f"indiquer le coefficient pour la ressource SAE{i}"))
    som_2 += note*coeff
    coeff_2 += coeff


moy_g = (som_1 + som_2) / (coeff_1 +coeff_2)
print(moy_g)
```
